```python
class MultiplicationTable:
    """
    MultiplicationTable class generates the multiplication table
    for a given number.

    Attributes:
        num (int): The number used to generate the multiplication table.
    """

    def __init__(self, num):
        self.num = num

    def multiplication_table(self):
        """
        Generates and prints the multiplication table from 1 to 10.

        Returns:
            None: This method prints the multiplication table.
        """
        for x in range(1, 11):
            print(f"{self.num} * {x} =", self.num * x)
```

```python
# Create an object
table = MultiplicationTable(9)

# Call the main method
table.multiplication_table()
```

<h1>Q2</h1>

```python
class TwinPrimes:
    """
    TwinPrimes class finds and prints all twin prime numbers
    less than a given limit.

    Attributes:
        limit (int): The upper limit for finding twin primes.
    """

    def __init__(self, limit):
        self.limit = limit

    def is_prime(self, num) -> bool:
        """
        Checks whether a number is prime.

        Parameters
        ----------
        num : int
            Number to test.

        Returns
        -------
        bool
            True if the number is prime, otherwise False.
        """
        if num < 2:
            return False

        for i in range(2, num):
            if num % i == 0:
                return False

        return True

    def twin_primes(self):
        """
        Prints all twin prime numbers less than the given limit.

        Returns
        -------
        None
            Prints the twin prime pairs.
        """
        for num in range(3, self.limit, 2):
            if self.is_prime(num) and self.is_prime(num + 2):
                print(f"({num}, {num + 2})")
```

```python
# Create an object
task = TwinPrimes(1000)

# Call the main method
task.twin_primes()
```

<h1>Q3</h1>

```python
class PrimeFactors:
    """
    PrimeFactors class finds the prime factors of a given number.

    Attributes:
        num (int): The number to factorize.
    """

    def __init__(self, num: int):
        self.num = num

    def primeFactors(self) -> list[int]:
        """
        Finds the prime factors of the number.

        Returns
        -------
        list[int]
            A list containing the prime factors of the number.
        """
        factors = []
        divisor = 2
        num = self.num

        while num > 1:
            while num % divisor == 0:
                factors.append(divisor)
                num //= divisor
            divisor += 1

        return factors
```

```python
prime = PrimeFactors(56)

prime.primeFactors()
```

<h1>Q4</h1>

```python
class DecToBin:
    """
    DecimalToBinary class converts a decimal number
    to its binary representation.

    Attributes:
        num (int): The decimal number to convert.
    """
    def __init__(self, num : int):
        self.num = num

    def decToBin(self) -> str:
        """
        Convert a decimal number to its binary representation.

        Parameters
        ----------
        num : int
            Decimal number to convert.

        Returns
        -------
        str
            Binary representation of the given decimal number.
        """
        binary = ""
        num = self.num

        while num > 0:
            binary = str(num % 2) + binary
            num //= 2

        return binary
```

```python
bin = DecToBin(11)

bin.decToBin()
```

```python
class PerfectNums:
    """
    PerfectNumbers class finds and prints all perfect numbers
    within a given range.

    Attributes:
        start (int): Starting value of the range.
        end (int): Ending value of the range.
    """
    def __init__(self, start: int, end: int):
        self.start = start
        self.end = end
    
    def perfectNums(self) -> None:
        """
        Print all perfect numbers within a given range.

        Parameters
        ----------
        start : int
            Starting value of the range.

        end : int
            Ending value of the range.

        Returns
        -------
        None
            This function prints the perfect numbers in the specified range.
        """
        for num in range(self.start, self.end + 1):
            total = 0

            for i in range(1, num):
                if num % i == 0:
                    total += i

            if total == num:
                print(num)
```

```python
perfect = PerfectNums(0, 100)

perfect.perfectNums()
```

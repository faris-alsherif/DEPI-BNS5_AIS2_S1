```python
def get_info(name :str, age : int):
    '''
        function for person info get the name and age and
        return back again to user
        
        Args:
            param_1 : name get the person name
            type of param_1 : str

            param_2 : age get the person age
            type of param_2 : int

            return : this function return the name and age
            type_ return : str

    '''
    return f"my name is {name}, my age is {age}"
```

```python
def str_to_lst(word = str) ->list[str]:
    '''
    Transform sentence of word to list of words.
        
    Parameters
    ----------
    word : str
    sentence of word.
        
    Returns
    -------
    list of String
    list contains the words of sentance.
    
    '''
    lst = word.split()
    return lst 
```

```python
str_to_lst("faris sherif taha")
```

```python
words = ["lol","anas","lol","hi","lol","hi"]
unique_words= set(words)
counts = {word: 0 for word in unique_words}
for i in unique_words:
    for j in words:
        if i == j:
            counts[i] += 1
```

```python
def count_words(*words: str) :
    unique_words= set(words)
    counts = {word: 0 for word in unique_words}
    for i in unique_words:
        for j in words:
            if i == j:
                counts[i] += 1
    return counts
```

```python
count_words("lol","anas","lol","hi","lol","hi")
```

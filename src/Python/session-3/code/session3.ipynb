```python
def get_info(name :str, age : int):
    '''
        function for person info get the name and age and
        return back again to user
        
        Args:
            param_1 : name get the person name
            type of param_1 : str

            param_2 : age get the person age
            type of param_2 : int

            return : this function return the name and age
            type_ return : str

    '''
    return f"my name is {name}, my age is {age}"
```

```python
get_info("faris", 21)
```

```python
get_info("Ahmed", 20)
```

```python
get_info(20, "Faris")
```

```python
get_info(age = 20, name = "Faris")
```

```python
def multiplicationTable() -> None:
    """
        Function that make multiplication table for the input number 
    """
    num = int(input("Enter a number:"))
    for x in range(1,11):
        print(f"{num} * {x} =",num*x)
```

<h1>Q2</h1>

```python
def is_prime(num) -> bool:
    """
    Check whether a number is prime.
    
    Parameters
    ----------
    num : int
    Number to test.
    
    Returns
    -------
    bool
    True if prime, otherwise False.
    """    
    if num < 2:
        return False
     
    for i in range(2,num):
        if num % i == 0:
            return False

    return True

def twinPrimes() -> None:
    """
    Print all twin primes less than 1000.
    """
    for num in range(3, 1000, 2):
        if is_prime(num) and is_prime(num + 2):
            print(f"({num}, {num + 2})")
```

```python
twinPrimes()
```

<h1>Q3</h1>

```python
def primeFactors(num: int) -> list[int]:
    """
    Find the prime factors of a number.

    Parameters
    ----------
    num : int
        Number to factorize.

    Returns
    -------
    list[int]
        A list containing the prime factors of the number.
    """
    factors = []
    divisor = 2

    while num > 1:
        while num % divisor == 0:
            factors.append(divisor)
            num //= divisor
        divisor += 1

    return factors
```

```python
primeFactors(56)
```

```python
def decToBin(num: int) -> str:
    """
    Convert a decimal number to its binary representation.

    Parameters
    ----------
    num : int
        Decimal number to convert.

    Returns
    -------
    str
        Binary representation of the given decimal number.
    """
    binary = ""

    while num > 0:
        binary = str(num % 2) + binary
        num //= 2

    return binary
```

```python
decToBin(11)
```

```python
def perfectNums(start: int, end: int) -> None:
    """
    Print all perfect numbers within a given range.

    Parameters
    ----------
    start : int
        Starting value of the range.

    end : int
        Ending value of the range.

    Returns
    -------
    None
        This function prints the perfect numbers in the specified range.
    """
    for num in range(start, end + 1):
        total = 0

        for i in range(1, num):
            if num % i == 0:
                total += i

        if total == num:
            print(num)


perfectNums(0, 100)
```

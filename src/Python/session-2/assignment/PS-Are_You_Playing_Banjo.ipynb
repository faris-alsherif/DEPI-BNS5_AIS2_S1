```python
def AreYouPlayingBingo(name):
    name = name.lower()
    if name.startswith('r'):
        name = name.capitalize()
        return name + " plays banjo"
    else:
        name = name.capitalize()
        return name + " does not play banjo"
```

```python
name = input("Enter your name: ")
AreYouPlayingBingo(name)
```

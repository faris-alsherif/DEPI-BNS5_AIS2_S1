```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic.csv")
df.head()
```

```python
df.drop(["PassengerId", "Name", "Ticket"], axis=1, inplace=True)
df.head()
```

```python
null = df.isnull().sum()
ratio = null / df.shape[0]
pd.DataFrame({"nulls_sum": null, "Ratio": ratio}).T
```

```python
df = df.dropna(subset= ['Embarked'])
```

```python
df = df.drop("Cabin", axis = 1)
```

```python
df['Age'] = df["Age"].fillna(df['Age'].median())
```

```python
null = df.isnull().sum()
ratio = null / df.shape[0]
pd.DataFrame({"nulls_sum": null, "Ratio": ratio}).T
```

```python
x = df.drop("Survived", axis=1)
y = df[["Survived"]]
```

```python
cols = ['Pclass', 'Survived', 'SibSp', 'Sex', 'Parch', 'Embarked']
df[cols] = df[cols].astype('category')
pd.DataFrame(df.dtypes).T
```

```python
nums_col = df.select_dtypes("number").columns
nums_col
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
# scaler.fit(x[nums_col])
x[nums_col] = scaler.fit_transform(x[nums_col])
x[nums_col]
```

```python
x.head()
```

```python
str_cols = ["Sex", "Embarked"]
```

```python
from category_encoders import OneHotEncoder
encoder = OneHotEncoder(cols = str_cols, drop_invariant= True)
x = encoder.fit_transform(x)
```

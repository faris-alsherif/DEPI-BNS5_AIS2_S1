<h1>Reverse Words in a String</h1>

```python
class Solution:
    def reverseWords(self, s: str) -> str:

        allowed = set(" 1234567890abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ")

        if not 1 <= len(s) <= 10000:
            raise ValueError("The length of s must be between 1 and 10000")
        elif not set(s).issubset(allowed):
            raise ValueError("s contains English letters (upper-case and lower-case), digits, and spaces ' '")
        else:
            reversed_words = ' '.join(s.split()[::-1])
            return reversed_words
```

```python
s1 = Solution()
s1.reverseWords("    a good   example    ")
```

<h1>Product of Array Except Self</h1>

```python
class Solution:
    def productExceptSelf(self, nums: list[int]) -> list[int]:
        n = len(nums)
        answer = [1] * n

        prefix = 1
        for i in range(n):
            answer[i] = prefix
            prefix *= nums[i]

        postfix = 1
        for i in range(n - 1, -1, -1):
            answer[i] *= postfix
            postfix *= nums[i]

        return answer
```

```python
s2 = Solution()
s2.productExceptSelf([-1,1,0,-3,3])
```

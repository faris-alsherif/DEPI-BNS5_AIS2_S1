<h1>Roman to Integer<h1>

```python
def romanToInt(s: str):
    integar = 0
    s = list(s)
    for i in range(0, len(s)-1):
        for j in range(i+1, len(s)):
            if s[i] == "I" and s[j] == "V":
                del(s[j])
                s[i] = "IV"
            elif s[i] == "I" and s[j] == "X":
                del(s[j])
                s[i] = "IX"
            elif s[i] == "X" and s[j] == "L":
                del(s[j])
                s[i] = "XL"
            elif s[i] == "X" and s[j] == "C":
                del(s[j])
                s[i] = "XC"
            elif s[i] == "C" and s[j] == "D":
                del(s[j])
                s[i] = "CD"
            elif s[i] == "C" and s[j] == "M":
                del(s[j])
                s[i] = "CM"
    for x in s:
        if x == "I":
            integar += 1
        elif x == "V":
            integar += 5
        elif x == "X":
            integar += 10
        elif x == "L":
            integar += 50
        elif x == "C":
            integar += 100
        elif x == "D":
            integar += 500
        elif x == "M":
            integar += 1000
        elif x == "IV":
                integar += 4
        elif x == "IX":
            integar += 9
        elif x == "XL":
            integar += 40
        elif x == "XC":
            integar += 90
        elif x == "CD":
            integar += 400
        elif x == "CM":
            integar += 900
    return integar
```

```python
romanToInt("MCMXCIV")
```

<h1>Search Insert Position</h1>

```python
def serchidx(nums: List, target: int):
    if not 1 <= len(nums) <= 10**4:
        raise ValueError("the length of list must be between 1 and 10000")
    elif not -10**4 <= target <= 10**4:
        raise ValueError("the value of target must be between -10000 and 10000")
    elif len(nums) != len(set(nums)):
        raise ValueError("the values of list must be distinct")

    for i in range(0, len(nums)):
        if not -10**4 <= nums[i] <= 10**4:
            raise ValueError(f"the num in index:{i} must be between -10000 and 10000")
        if nums[i] == target:
            return i
        else:
            if nums[len(nums)-1] < target:
                return len(nums)
            elif nums[i] < target < nums[i+1]:
                return i+1
            elif nums[0] > target:
                return 0
```

```python
serchidx([3,4,9,10], 5)
```

<h1>Basic Calculator II</h1>

```python
def calculate(s):
    stack = []
    num = 0
    op = '+'

    for i, ch in enumerate(s):
        if ch.isdigit():
            num = num * 10 + int(ch)

        if ch in '+-*/' or i == len(s) - 1:
            if op == '+':
                stack.append(num)

            elif op == '-':
                stack.append(-num)

            elif op == '*':
                stack.append(stack.pop() * num)

            elif op == '/':
                stack.append(int(stack.pop() / num))

            op = ch
            num = 0

    return sum(stack)
```

```python
calculate("2+5*3")
```
